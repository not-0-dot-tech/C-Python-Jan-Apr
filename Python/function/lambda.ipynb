```python
'''
def sum(a, b):
    return a + b

sum(1, 2)
'''
```

```python
sum = lambda a, b : a + b

print(sum(1, 2))
```

```python
def creator_func(n):
    return lambda a : a * n

doubler_func = creator_func(2) # lambda a : a * 2
tripler_func = creator_func(3) # lambda a : a * 3

print(doubler_func(5)) # 10
print(tripler_func(5)) # 15
```

```python
another_creator_func = lambda n : lambda a : a * n
doubler_func = another_creator_func(2) # lambda a : a * 2
tripler_func = another_creator_func(3) # lambda a : a * 3

print(doubler_func(5)) # 10
print(tripler_func(5)) # 15
```
