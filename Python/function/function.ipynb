```python
def greet(): # defining a function
    print("Hello!")

greet() # calling the function
    
```

```python
def greet(name): # defining a function with a parameter
    print(f"Hello, {name}!")

greet("Alice") # calling the function with an argument
```

```python
def add(x, y): # defining a function with two parameters
    return x + y

result = add(5, 3) # calling the function with arguments
print(f"The sum is: {result}") # printing the result
```
