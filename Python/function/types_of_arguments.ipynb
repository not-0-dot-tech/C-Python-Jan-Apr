### Positional Argument

```python
def relation(elder, younger):
    print(f"{elder} is elder than {younger}.")


relation('Avik', 'Alik')
relation('Alik', 'Avik')
```

### Variable Positional Argument

```python
def func(*arg):
    print(f"{arg[0]} is elder.")
    print(f"{arg[1]} is a younger.")

func('Avik', 'Alik')
```

```python
def find_sum(*values):
    print(f"Sum of {values} is {sum(values)}")

find_sum(1, 2)
find_sum(1, 2, 3)
find_sum(1, 2, 3, 4)
find_sum(1, 2, 3, 4, 5)
```

```python
def magic_number(actual_number, secret_number = 5): # dafault value is 5
    print(f"Magic number is {secret_number + actual_number ** 2}.")

magic_number(2, 3) # 3 will be used for secret_number
magic_number(3) # default value 5 will be used for secret_number
```

### Keyword arguments

```python
def relation(elder, younger):
    print(f"{elder} is elder than {younger}.")

relation('Avik', 'Alik') # positional argument
relation(younger='Alik', elder='Avik') # keyword argument
relation(elder='Avik', younger='Alik') # keyword argument
```

## Variable Keyword Argument

```python
def func(**kwargs):
    print(f"Argument is {kwargs}")

func(name='Avik', age=20, city='Kolkata')
func(name='Avik', age=20)
```

```python
def func(**kwargs):
    print(f"{kwargs['elder']} is elder.")
    print(f"{kwargs['younger']} is a younger.")

func(elder='Avik', younger='Alik')
```

### Positional Only Argument

```python
def sum(a, b, /): # positional-only arguments
    print(f"Sum of {a} and {b} is {a + b}.")
```

```python
sum(1, 2) # positional arguments
```

```python
sum(a=1, b=2) # TypeError
```

```python
def sum(a, b, /, c, d):
    print(f"Sum of {a}, {b}, {c} and {d} is {a + b + c + d}.")
```

```python
sum(1, 2, 3, 4) # positional arguments
```

```python
sum(a=1, b=2, c=3, d=4) # TypeError
```

```python
sum(1, 2, d=4, c=3)
```

### Keyword Only Argument

```python
def sum(*, a, b): # keyword-only arguments
    print(f"Sum of {a} and {b} is {a + b}.")
```

```python
sum(a=1, b=2) # keyword arguments
```

```python
sum(1, 2) # TypeError
```

### Combine positional only and keyword only arguments

```python
def sum(a, b, /, *, c, d): # positional and keyword arguments
    print(f"Sum of {a}, {b}, {c} and {d} is {a + b + c + d}.")

sum(1, 2, c=3, d=4) # positional and keyword arguments
```
