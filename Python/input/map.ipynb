```python
def find_length(s):
    return len(s)

names = ('Avik', 'Prakash', 'Jyoti', 'Sourabh')

l = list(map(find_length, names))

print(l)
```

```python
find_length = lambda s: len(s)

list(map(find_length, ('Avik', 'Prakash', 'Jyoti', 'Sourabh')))
```
